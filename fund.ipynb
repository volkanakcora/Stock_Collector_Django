```python
import http.client
import json
import pandas as pd
from time import sleep

headers = {
    'x-rapidapi-key': "c3c297e087mshf0c09b3c9068e2fp1ae78fjsnf0b02d992e8a",
    'x-rapidapi-host': "tefas-api.p.rapidapi.com"
}

conn = http.client.HTTPSConnection("tefas-api.p.rapidapi.com")

# 1️⃣ Tüm fon listesini çek
conn.request("GET", "/api/v1/funds", headers=headers)
res = conn.getresponse()
data = json.loads(res.read().decode("utf-8"))

funds = data.get("data", [])
print(f"Toplam {len(funds)} fon bulundu.")
print("İlk 5 fon:", funds[:5])

# 2️⃣ Her fon için geçmiş verileri çek
all_fund_data = []
for fund in funds:
    code = fund["key"]
    print(f"Fon çekiliyor: {code}")
    conn.request("GET", f"/api/v1/funds/historical?fundCode={code}", headers=headers)
    res = conn.getresponse()
    fund_data = json.loads(res.read().decode("utf-8"))
    records = fund_data.get("data", [])
    if records:
        for r in records:
            r["fund_code"] = code
        all_fund_data.extend(records)
    sleep(0.2)  # RapidAPI rate limit için

# 3️⃣ DataFrame’e çevir
df = pd.DataFrame(all_fund_data)
print(df.head())

# 4️⃣ Örnek: Fon getirilerini hesaplamak
df["return_percent"] = df["price"].pct_change() * 100
```

```python
import http.client
import json
import pandas as pd
from datetime import datetime, timedelta

# RapidAPI bağlantı bilgileri
headers = {
    'x-rapidapi-key': "c3c297e087mshf0c09b3c9068e2fp1ae78fjsnf0b02d992e8a",
    'x-rapidapi-host': "tefas-api.p.rapidapi.com"
}
conn = http.client.HTTPSConnection("tefas-api.p.rapidapi.com")

# 📅 Bugünün tarihi
today = datetime.today()

# 🔁 Hesaplanacak periyotlar
periods = {
    "1M": today - timedelta(days=30),
    "3M": today - timedelta(days=90),
    "6M": today - timedelta(days=180),
    "1Y": today - timedelta(days=365)
}

all_data = []

for label, start_date in periods.items():
    start = start_date.strftime("%d.%m.%Y")
    end = today.strftime("%d.%m.%Y")
    print(f"📊 {label} getirileri alınıyor: {start} - {end}")
    
    conn.request("GET", f"/api/v1/funds/returns?fundType=1&startDate={start}&endDate={end}", headers=headers)
    res = conn.getresponse()
    data = json.loads(res.read().decode("utf-8"))
    
    records = data.get("data", [])
    if not isinstance(records, list):
        records = [records] if records else []
    
    df = pd.DataFrame(records)
    df["period"] = label  # hangi dönem olduğunu işaretle
    all_data.append(df)

# 🔗 Tüm periyotları birleştir
result = pd.concat(all_data, ignore_index=True)

print(result.head())
```

```python
import yfinance as yf
import pandas as pd
from datetime import datetime

def get_stock_data_safe(ticker, start_date, end_date, retries=3):
    """
    Production-ready yfinance veri çekme fonksiyonu
    
    Args:
        ticker: Hisse sembolü (AAPL, GARAN.IS, etc.)
        start_date: YYYY-MM-DD formatında
        end_date: YYYY-MM-DD formatında
        retries: Deneme sayısı
    
    Returns:
        DataFrame veya None
    """
    for attempt in range(retries):
        try:
            # Ticker objesi oluştur
            stock = yf.Ticker(ticker)
            
            # Veri çek
            df = stock.history(start=start_date, end=end_date)
            
            if df.empty:
                print(f"⚠️ {ticker}: Veri boş (delisted olabilir)")
                return None
            
            # Index'i sütun yap
            df = df.reset_index()
            
            # Ticker kolonu ekle
            df['ticker'] = ticker
            
            print(f"✅ {ticker}: {len(df)} satır veri")
            return df
            
        except Exception as e:
            if attempt < retries - 1:
                print(f"⚠️ {ticker}: Hata (Deneme {attempt + 1}/{retries}), tekrar deneniyor...")
                continue
            else:
                print(f"❌ {ticker}: {str(e)}")
                return None
    
    return None

# TEST
print("🧪 Production-ready fonksiyon testi:")
print("=" * 60)

result = get_stock_data_safe('AAPL', '2024-10-01', '2024-10-25')
if result is not None:
    print("\nSONUÇ:")
    print(result.head())
    print(f"\nKolonlar: {result.columns.tolist()}")
```
